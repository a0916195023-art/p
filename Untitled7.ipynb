```python
# 1. 冒險者登入
name = input("請輸入你的冒險者名字：")

# 讓玩家從三項指定裝備中挑選
print("\n--- 🔧 選擇你的傳奇裝備 ---")
print("1. 【黃金鍵盤】（技能：開局自動獲得 1 勝，搶佔先機！）")
print("2. 【魔法魔杖】（技能：平手時會施展法術，直接算你贏！）")
print("3. 【幸運硬幣】（技能：當電腦拿到 4 勝時，有機率強行重來該回合！）")

while True:


    eq_choice = input("請輸入數字選擇裝備 (1-3)：")
    if eq_choice == "1":
        equipment = "黃金鍵盤"
        break
    elif eq_choice == "2":
        equipment = "魔法魔杖"
        break
    elif eq_choice == "3":
        equipment = "幸運硬幣"
        break
    else:
        print("⚠️ 選擇錯誤，請輸入 1, 2 或 3。")

# 印出前置劇情
print("\n=== 系統廣播 ===")
print(f"歡迎勇敢的冒險者【{name}】！")
print(f"你已經裝備了具有神祕力量的【{equipment}】，準備好迎接挑戰！")

# 最外層的「再玩一次」大迴圈
while True:
    print("\n\n=== ✊✋✌️ 剪刀石頭布 終極大對決 ===")
    print("規則：先獲得 5 勝的一方就是最終贏家！\n")

    # 戰鬥前置作業
    # 裝備功能 1：【黃金鍵盤】開局直接獲得 1 勝！
    if equipment == "黃金鍵盤":
        player_wins = 1
        print("⚡ 【黃金鍵盤】發動！強大的打字速度讓你在開局就直接獲得 1 勝！")
    else:
        player_wins = 0

    computer_wins = 0
    round_num = 1
    options = ["剪刀", "石頭", "布"]

    # 2. 迴圈條件設定
    while player_wins < 5 and computer_wins < 5:
        print(f"--- 第 {round_num} 回合 ---")
        print(f"🏆 目前戰況 - 你：{player_wins} 勝 | 電腦：{computer_wins} 勝")

        # 3. 玩家出拳
        player_choice = input("請出拳 (輸入 剪刀、石頭 或 布)：")

        # 檢查玩家是否輸入錯誤
        if player_choice != "剪刀" and player_choice != "石頭" and player_choice != "布":
            print("⚠️ 哎呀！指令錯誤，請輸入 剪刀、石頭 或 布。\n")
            continue

        # 4. 電腦隨機出拳
        computer_choice = random.choice(options)
        print(f"🤖 電腦出了：{computer_choice}")

        # 5. 勝負判斷
        if player_choice == computer_choice:
            # 裝備功能 2：【魔法魔杖】平手強行判定為玩家勝利！
            if equipment == "魔法魔杖":
                print("🔮 【魔法魔杖】閃爍！咒語將平手局勢扭轉，這回合算你贏！\n")
                player_wins = player_wins + 1
            else:
                print("平手！沒有人得分。\n")

        elif (player_choice == "剪刀" and computer_choice == "布") or \
             (player_choice == "石頭" and computer_choice == "剪刀") or \
             (player_choice == "布" and computer_choice == "石頭"):
            print("🎉 這回合你贏了！\n")
            player_wins = player_wins + 1
        else:
            # 裝備功能 3：【幸運硬幣】在電腦即將贏得比賽（4 勝）時，有 50% 機率逆天改命重來！
            if equipment == "幸運硬幣" and computer_wins == 4 and random.choice([True, False]):
                print("🪙 【幸運硬幣】在空中翻轉！幸運女神對你微笑，這回合電腦的勝利被抹消了！\n")
                # 不幫電腦加分，直接進行下一回合
            else:
                print("💥 這回合電腦贏了！\n")
                computer_wins = computer_wins + 1

        round_num = round_num + 1

    # 6. 最終結算
    print("=== 🌟 最終結果 🌟 ===")
    if player_wins >= 5: # 用 >= 避免黃金鍵盤或魔杖溢出分數導致沒判斷到
        print(f"🏆 恭喜你以 {player_wins} : {computer_wins} 擊敗了電腦，成為剪刀石頭布大師！")
    else:
        print(f"💻 很惜，電腦以 {computer_wins} : {player_wins} 獲得了最終勝利，下次再來挑戰吧！")

    # 詢問是否再玩一次
    print("\n===============================")
    play_again = input("冒險者，想要再挑戰一次嗎？(y/n)：").lower()

    if play_again != 'y':
        print(f"\n謝謝遊玩！期待【{name}】下一次的冒險！✨")
        break
```
